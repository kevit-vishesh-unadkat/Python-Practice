```python
# if statement 

age = int(input("enter your age"))

if age > 18 :
    print("you can vote")
```

```python
# if statememt with else

age = int(input("enter your age"))

if age > 18:
    print("you can vote")

else:
    print("you can not vote")
```

```python
# if statement with elif and else

marks=int(input("enter your mark"))

if marks>90:
    print('A')
elif marks>80:
    print('B')
elif marks>70:
    print('C')
elif marks>60:
    print('D')
else:
    print('F')
```

```python
# conditional statement with membership operator

email=input("enter your email")

if "@gmail.com" not in email:
    print("Invalid email")

else:
    print("valid email")
```
