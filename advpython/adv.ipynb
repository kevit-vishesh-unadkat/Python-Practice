### Decorators

A decorator in python is a function that receives another function as input and adds some functionality(decoration) to and it and returns it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python
- `Built in decorators` like `@staticmethod`, `@classmethod`, `@abstractmethod` and `@property` etc
- `User defined decorators` that we programmers can create according to our needs

```python
# in python ->python functions are a first class citizens

# what is the meaning of first class citizens

# ->so in python they can be used just like numbers, string or any other variables

# Assign functions to variables.
# Pass them as arguments to other functions.
# Return them from functions.
# Store them in data structures such as lists or dictionaries.
```

```python
def my_decorator(func):
    def wrapper():
        print("*********************")
        func()
        print("*********************")
    return wrapper 

def hello():
    print("hello")   

a=my_decorator(hello)
a()
```

```python
def greet():
    print('Hello From Vishesh')

greeting=greet

greeting()

# it does not strore the function
# it does store the refernce of a fumction
```

```python
# passing a function to the onother function

def greet():
    print("Hello from visshesh")

def execute_task(func):
    print("starting task")
    func()
    print('ending task')

execute_task(greet)
```
