```python
# generator

# generator is a speacial  kind of iterator

# yield keyword

def generate_number():
    yield 10
    yield 20
    yield 30

generate_number()
# this will return the only generator object
number=generate_number()
print(next(number))
print(next(number))
print(next(number))
print(next(number))

```

```python
# we can also aply a loop on it 

def generate_number():
    yield 10
    yield 20
    yield 30

number=generate_number()

for i in number:
    print(i)
```

```python
def read_logs(filename):

    with open(filename,"r") as file:
        for line in file:
            yield line.strip()

log=read_logs("application.log")
print(next(log))

print(next(log))
print(next(log))


# for i in read_logs("application.log"):
#     print(i)
```

generator expression

```python
# like list comprehension
# in generator also we have generator  expression


square=[x*x for x in range(1,6)]
print(square)
print(type(square))
```

```python
square=(x*x for x in range(10))
print(type(square))
print(next(square))
print(next(square))
print(next(square))

# from this we can start a loop then it not start from the first 
# it will start from 9

for i in square:
    print(i)

# so generator are statefull
```

```python
# generator methods

# generator provides some several important methods

# __next__()
# send()
# throw()
# close()

def number():
    yield 10
    yield 20

g=number()

print(g.__next__())
print(g.__next__())
```
