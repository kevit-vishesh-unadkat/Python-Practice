Regular Expression in python 

support through the built-in re module

```python
import re

text = "python is powerful"

result=re.search(r"python",text)

print(result)

answer=re.findall(r"python",text)
print(answer)
# findall return the python list 
# and search will return the indexes


print(result.group())
```

```python
# Dot . any character except new line

re.findall(r"c.t","car cat ,cot ,bot")
```

```python
text = "python regular expression"

re.findall(r"[aeiou]",text)


# re.findall(r"[AEIOU]",text)   this will return the empty list
```

```python
# Negated Character classs

re.findall(r"[^0-9]","A123b567")
```

```python
# Quantifiers

re.findall(r"ab*","a ab abb abbbbb abbbbbbbbbbb")
```

```python
# Anchors


# Anchors doesn't consume character

re.search(r"^python","python is great")
```

```python
re.search(r"^Pyhon","python is great") 
# this will not return anything
```

```python
# Groups (...)

text = "cat dog bird"

re.findall(r"(cat|dog)",text)
```

```python
# Multiple Groups

text="vishesh - 9106891863"

match =re.search(r"(\w+)\s*-\s*(\d+)",text)

print(match)

print(match.group(0))
print(match.group(1))
print(match.group(2))
# print(match.group(3))   this will throw the IndexError Exception

```

```python
# Named Groups

pattern = r"""
(?P<name>\w+)
\s*-\s*
(?P<phone>\d+)
"""

match = re.search(pattern, "vishesh - 9106891863" , re.VERBOSE)

print(match.group("name"))
print(match.group("phone"))
```

```python
text="Order ID : 12345"

match=re.search(r"\d+",text)

print(match.start())
print(match.end())
```

```python
text= "Order ID : 12345678"

match = re.search(r"\d+",text)

# print(match.endpos)
print(match.start())
print(match.end())
print(match.span())
```

```python
text= "Order ID : 12345678"

match = re.search(r"\D",text)

# print(match.endpos)
print(match.start())
print(match.end())
print(match.span())
```

```python
text= "Order ID : 12345678"

match = re.search(r"\d+",text)

# print(match.endpos)
print(match.start())
print(match.end())
print(match.span())

print(match.string)
print(match.re)
```

```python
text = """ERROR database failed
INFO server started
ERROR timeout"""

print(re.findall(r"^ERROR.*$", text, re.MULTILINE))
```

```python
import re

data = """1. Apple
2. Banana
Not a fruit
3. Cherry"""

# WITHOUT re.MULTILINE (Only looks at the very first character of 'data')
print(re.findall(r"^\d\.", data))
# Output: ['1.']

# WITH re.MULTILINE (Looks at the start of every line)
print(re.findall(r"^\d\.", data, flags=re.MULTILINE))
# Output: ['1.', '2.', '3.']
```

```python
import re

phone_regex = re.compile(r"""
    ^                      # Match start of string
    (
        \(\d{3}\)          # Matches an area code in parentheses: e.g., (123)
        |                  # OR
        \d{3}              # Matches a plain 3-digit area code: e.g., 123
    )
    [-\s]?                 # Optional separator (dash or space)
    \d{3}                  # 3-digit prefix
    -                      # Match a literal dash
    \d{4}                  # 4-digit line number
    $                      # Match end of string
""", re.VERBOSE)

# Both patterns work exactly the same way:
print(phone_regex.match("(123) 456-7890")) # Matches successfully
```
