```python
# catching multiple exception

try:
    f=open('sample.txt','r')
    print(f.read())

    # print(m)
    # l=[1,2,3]
    # l[100]
    print(5/0)

except NameError:
    print("you not variable")

except IndexError:
    print("you can access the index witch not prsent")

except ZeroDivisionError:
    print("you can not divide any number with zero ")

except Exception as e:
    print(e)
```

```python
# except and else block example

try:
    f=open('sample.txt','r')

except FileNotFoundError:
    print("file nai mili")

except Exception as e:
    print("bhai kuch to lafda hai")

else:
    print(f.read())
```

```python
# else , except and finally block

try:
    f=open('sample1.txt','r')

except FileNotFoundError:
    print("file nai mili")

except Exception as e:
    print("bhai kuch to lafda hai")

else:
    print(f.read())

finally:
    print("ye to excute hoga hi")
```

```python
#  raise an exception

class Bank:

    def __init__(self,balence):
        self.balence=balence

    def withdraw(self,amount):
        if amount<0:
            raise Exception('amount cannot be -ve')
        elif self.balence<amount:
            raise Exception('paise nai hai tere pass')
        else:
            self.balence=self.balence-amount


obj=Bank(10000)

try:
    obj.withdraw(15000)
except Exception as e:
    print(e)
else:
    print(obj.balence)
```

```python
class MyException(Exception):
  def __init__(self,message):
    print(message)

class Bank:

  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise MyException('amount cannot be -ve')
    if self.balance < amount:
      raise MyException('paise nai hai tere paas')
    self.balance = self.balance - amount

obj = Bank(10000)
try:
  obj.withdraw(5000)
except MyException as e:
  pass
else:
  print(obj.balance)
```

```python
class SecurityError(Exception):

  def __init__(self,message):
    print(message)

  def logout(self):
    print('logout')

class Google:

  def __init__(self,name,email,password,device):
    self.name = name
    self.email = email
    self.password = password
    self.device = device

  def login(self,email,password,device):
    if device != self.device:
      raise SecurityError('bhai teri to lag gayi')
    if email == self.email and password == self.password:
      print('welcome')
    else:
      print('login error')



obj = Google('nitish','nitish@gmail.com','1234','android')

try:
  obj.login('nitish@gmail.com','1234','windows')
except SecurityError as e:
  e.logout()
else:
  print(obj.name)
finally:
  print('database connection closed')
```
