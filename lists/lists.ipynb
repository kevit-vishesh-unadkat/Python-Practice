```python
# sorted

# it creates a new sorted list

num=[5,4,3,2,1]

new_num=sorted(num)
print(new_num)
print(num)
```

```python
# iterating over a list

products = ["Laptop", "Mouse", "Keyboard"]

for product in products:
    print(product)
```

```python
# enumerate over the list

products = ["Laptop", "Mouse", "Keyboard"]

for index,product in enumerate(products):
    print(index,product)
```

```python
# zip

names=["vishesh","yug","priyank"]
scores=[90,85,95]
for name,score in zip(names,scores):
    print(name,score)
```
