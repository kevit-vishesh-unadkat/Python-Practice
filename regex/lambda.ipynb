lambda with sorted()

```python
products = [
    {"name": "Laptop", "price": 75000},
    {"name": "Mouse", "price": 1200},
    {"name": "Keyboard", "price": 2500},
    {"name": "Monitor", "price": 18000}
]
```

```python
products.sort(key=lambda product : product["price"])
print(products)
```

```python
# sorting in decending order

products.sort(key=lambda product : product['price'],reverse=True)
print(products)
```

```python
employees.sort(key=lambda employee : (employee['salary'],employee['experience']))
print(employees)
```

lambda with map()

```python
number = [1,2,3,4]

result= map(lambda num : num * 2 , number)

print(result)   
# this will only return a map object not return the actual value after transformation 
# result is an iterator
print(number)
print(list(result))
# we need to convert this into list and return or print it
```

```python
# real world example

prices = [1000, 2500, 5000, 10000]

updated_price=list(map(lambda price : price * 1.10,prices))

print(updated_price)
```

map() with a string

```python
users=["vishesh","nirmit","yug"]

update_user=list(map(lambda user: user.upper(),users))

print(update_user)
```

map() with multiple list

```python
prices = [100,200, 300]
quantitys= [2,5,7]

total=list(map(lambda price,quantity : price * quantity
           ,prices
           ,quantitys))

print(total)
```

basic filter() with lambda

```python
numbers=[10,15,20,25,30]

even_number=filter(lambda num : num%2==0,numbers)

print(even_number)

# this will also return a filter iterator 
# not actual result 
# so we need to convert them into list

print(list(even_number))
```

```python
# real world example for filter() with lambda

users = [
    {"name": "Amit", "active": True},
    {"name": "Raj", "active": False},
    {"name": "Neha", "active": True},
]


active_users=list(filter(lambda user: user['active'],users))

print(active_users)
```

any() -Higher order thinking

any() itself does not take a function

it receives an iterable and return True if at least one element is truthy

```python
scores=[35,42,76,51]

result=any(score>=70 for score in scores)

print(result)
```

can also combine with map()

```python
result = any(
    map(lambda score : score>=70,scores)
)

print(result)
```

all() ->returns the true when every element is truthy.

for empty iterable it returns True

```python
password = "Python@12345"

checks = [
    len(password) >= 8,
    any(c.isupper() for c in password),
    any(c.isdigit() for c in password),
    any(c in "@#$%" for c in password)
]

print(all(checks))
```

min() with lambda

```python
products = [
    {"name": "Laptop", "price": 75000},
    {"name": "Mouse", "price": 1200},
    {"name": "Monitor", "price": 18000}
]


cheapest=min(
    products,
    key=lambda product : product['price']
)

print(cheapest)
```

zip() + lambda + map()

```python
products = ["Laptop", "Mouse", "Keyboard"]
prices = [75000, 1200, 2500]


products_data= map(
    lambda name_price : {
        "name":name_price[0],
        "price":name_price[1]
    },
    zip(products,prices)
)

print(list(products_data))
```

lambda + filter() + map()

```python
orders = [
    {"customer": "Amit", "amount": 500},
    {"customer": "Raj", "amount": 2500},
    {"customer": "Neha", "amount": 4000},
    {"customer": "Jay", "amount": 800}
]

large_order= list(filter(lambda filter : filter['amount']>1000,orders))
print(large_order)

customers=list(map(lambda order : order['customer'],large_order))
print(customers)
```

```python
# real world example 

employees = [
    {"name": "Amit", "salary": 45000, "active": True},
    {"name": "Raj", "salary": 70000, "active": True},
    {"name": "Neha", "salary": 50000, "active": False},
    {"name": "Jay", "salary": 80000, "active": True}
]

high_paid_active= list(filter(lambda employee : employee['salary']>50000 and employee['active'],employees))

print(high_paid_active)
```

```python
names=list(map(lambda employee : employee['name'],high_paid_active))

print(names)
```

reduce()

```python
from functools import reduce

number=[10,20,30]

sum=reduce(lambda a,b:a+b,number)

print(sum)
```
