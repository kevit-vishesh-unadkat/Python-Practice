```python
# find()
# find the first occurrence
# it will return the index of first occurrence of that particuler .
# if not found then give output as -1.
email="vishesh@gmail.com"

print(email.find("@"))
```

```python
# real world use case

email=input("enter your email")

position=email.find("@")

username=email[:position]

print(username)
```
